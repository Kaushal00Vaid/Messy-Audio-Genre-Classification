```python
import numpy as np
from sklearn.model_selection import train_test_split

# 1. Generate dummy recipes (10 genres, 100 recipes each)
genres = ["blues", "classical", "country", "disco", "hiphop", "jazz", "metal", "pop", "reggae", "rock"]
all_recipes = []

for genre in genres:
    for i in range(100):
        all_recipes.append({"genre": genre, "file_id": f"{genre}_{i}"})

print(f"Total recipes generated: {len(all_recipes)}")

# 2. Split the dataset
train_recipes, val_recipes = train_test_split(
    all_recipes, 
    test_size=0.2, 
    shuffle=True, 
    random_state=42
)

print(f"Number of recipes in val_recipes: {len(val_recipes)}")
```

```python
# Simulating the librosa load process with dummy numpy arrays
# sr=16000, duration=10 means 160,000 samples per array

# 4 dummy stems
vocals = np.zeros(160000)
drums = np.zeros(160000)
bass = np.zeros(160000)
other = np.zeros(160000)

# 1 dummy noise file
noise = np.zeros(160000)

# Summing the 4 stems
mix_array = vocals + drums + bass + other

# Adding scaled noise
intensity_weight = 0.2
final_mix = mix_array + (noise * intensity_weight)

print(f"Shape of the final mix array: {final_mix.shape}")
```

```python
import torch
from transformers import AutoFeatureExtractor

# 1. Initialize the Hugging Face feature extractor
extractor = AutoFeatureExtractor.from_pretrained("MIT/ast-finetuned-audioset-10-10-0.4593")

# 2. Create the dummy audio mix
mix = np.ones(160000)

# 3. Process the audio
inputs = extractor(mix, sampling_rate=16000, return_tensors="pt")

# 4. Extract and squeeze the tensor
spectrogram_tensor = inputs["input_values"].squeeze(0)

print(f"Exact shape of the PyTorch tensor: {list(spectrogram_tensor.shape)}")
```

```python
from transformers import ASTForAudioClassification

# 1. Initialize the AST model with a modified classification head
model = ASTForAudioClassification.from_pretrained(
    "MIT/ast-finetuned-audioset-10-10-0.4593", 
    num_labels=10, 
    ignore_mismatched_sizes=True
)

# 2. Calculate trainable parameters
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"Number of trainable parameters: {trainable_params}")
```

```python
# 1. Create the dummy test array
y_test = np.array([-0.85, 0.40, 0.20, -0.10])

# 2. Apply the normalization formula
y_normalized = y_test / (np.max(np.abs(y_test)) + 1e-9)

# 3. Print the value at index 0 rounded to 3 decimal places
print(f"Normalized value at index 0: {y_normalized[0]:.3f}")
```
